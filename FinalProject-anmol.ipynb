```python
# BAMS 506 2025W1 BA2 - Optimal Decision Making I
# Final Project
# Group Members:
# Hans
# Model Formulations
# Baseline version (Part A)
# The general idea is to minimize the cost of providing food to children while still meeting their baseline nutritional requirements.

# Let:
# Xi = # of servings of a particular food {i = [1, 2, .., 49]}
# pi = price per serving of a particular food
# mj = variable unit of a particular nutrient {j = [1, 2, … 10]}

# Decision Variables:
# Per the food_data.xlsx table, let the following decision variables equal
# x1 = # of servings of Almonds
# x2 = # of servings of Apple Raw with skin
# …
# x49 = # of servings of White Tuna in Water

# Objective Function:
# Min  Pi*Xi = Sigma[i= 1 to 49](pi*xi) = p1*x1 +p2*x2 + … + p49*x49 = 0.5*x1 + 0.24*x2 + … + 0.69*x49

# Where:
# p1 = price per serving of Almonds
# p2 = price per serving of Apple Raw with skin
# …
# p49 = price per serving of White Tuna in Water

# Constraints:
# Idea behind constraint formulas is to set the upper and lower bound for each nutrient, from a per food serving perspective
# Nutrient Constraints
# Sigma[j = 1 to 10]Sigma[i = 1 to 49]{mj*xi} >= mjMin
# Sigma[j = 1 to 10]Sigma[i = 1 to 49]{mj*xi} <= mjMax

# Written more explicitly:
# m1Min   164*x1 + 81.4*x2 + … + 115.6*x3 >= 1800
# m1Max   164*x1 + 81.4*x2 + … + 115.6*x3 <= 2400
# m2Min   14.2*x1 + 0.5*x2 + … + 2.1*x3 >= 60
# m2Max   14.2*x1 + 0.5*x2 + … + 2.1*x3 <= 95
# …
# m10Min  1.1*x1 + 0.2*x2 + … + 0.5*x3 >= 8
# m10Max  1.1*x1 + 0.2*x2 + … + 0.5*x3 <= 40

# Where:
# m1 = calories measured in kcal
# m2 = fat measured in grams
# …
# m10 = iron measured in milligrams
# Balanced Diet Constraints:
# No single food item should make up more than 30% of total calories OR protein intake

# xi * m1 <=  (0.30) * Sigma[i = 1 to 49](xi * m1) + M * (yi),
# xi * m6 <=  (0.30) * Sigma[i = 1 to 49](xi * m6) + M * (1 - yi)

# Where:
# yi = [0, 1], binary activator
# M = “large constant” to make the value purposely high, in excess of any particular possible max nutrient for a given food, for example M = 10,000

# Non-Negativity
# xi >= 0
```

```python
#GenAI Disclaimer: Code was co-created with GenAI (GitHub Copilot)
# Part D — 5-Day plan (relaxed 30% cap by default), exports Week_Plan and Costs to Excel

import os
import pandas as pd
import gurobipy as gp
from gurobipy import GRB

# If an updated Excel (Food_Data (With Extension).xlsx) exists use it, otherwise fall back to in-notebook lists
EXCEL_PATH = "Food_Data (With Extension).xlsx"
# Exact nutrition column names as provided by the user
NUTRITION_COLS = [
    'Calories (kcal)', 'Fat (g)', 'Sodium (mg)', 'Carbs (g)', 'Fiber (g)',
    'Protein (g)', 'Vitamin A (IU)', 'Vitamin C (mg)', 'Calcium (mg)', 'Iron (mg)'
]
try:
    if os.path.exists(EXCEL_PATH):
        df_food = pd.read_excel(EXCEL_PATH, sheet_name=0)
        # Food name column (attempt common names)
        if 'Food' in df_food.columns:
            foods = df_food['Food'].astype(str).tolist()
        elif 'Name' in df_food.columns:
            foods = df_food['Name'].astype(str).tolist()
        else:
            # fallback to first column
            foods = df_food.iloc[:, 0].astype(str).tolist()

        # Read nutrition columns using the exact headers provided
        missing = [c for c in NUTRITION_COLS if c not in df_food.columns]
        if missing:
            raise KeyError(f"Missing nutrition columns in Excel: {missing}")
        nutrition_data = df_food.loc[:, NUTRITION_COLS].values.tolist()

        # Cost column
        if 'Cost' in df_food.columns:
            costs = df_food['Cost'].astype(float).tolist()
        elif 'Price' in df_food.columns:
            costs = df_food['Price'].astype(float).tolist()
        else:
            raise KeyError("Cost column not found in Excel (expected 'Cost' or 'Price').")

        # optional category column name fallback
        if 'Category' in df_food.columns:
            food_group_labels = df_food['Category'].astype(str).tolist()
        else:
            food_group_labels = None
    else:
        raise FileNotFoundError
except Exception:
    # fall back to variables defined earlier in notebook (these names exist in earlier cells)
    try:
        foods, nutrition_data, costs, food_group_labels  # noqa: F821
    except Exception:
        raise RuntimeError("No food data available. Add 'Food_Data (With Extension).xlsx' or ensure earlier cells run.")

# prepare food groups mapping (if food_group_labels provided)
food_groups = {}
if 'food_group_labels' in globals() and food_group_labels is not None:
    for idx, label in enumerate(food_group_labels):
        food_groups.setdefault(label, []).append(idx)
else:
    # if no labels available, put everything into 'Other'
    food_groups = {"Other": list(range(len(foods)))}

# PARAMETERS - tweak as needed
DAYS = 5  # Mon-Fri
NFOODS = len(foods)
MIN_DISTINCT_ITEMS_DAY = 3        # minimum different foods per day
MAX_REPEAT_PER_WEEK = 4        # max appearances per week per food
MIN_SERVING_IF_USED = 0.5         # minimum serving if a food is used that day (helps avoid numerical tiny values)
BIG_M = 20.0                      # large M for linking
ENFORCE_30_PCT_CAP = False        # toggle to enforce per-day 30% cap (may become infeasible)
OUTPUT_XLSX = "Starlight_Optimizer_Output.xlsx"

# Build model
model = gp.Model("PartD_5Day_Relaxed30pct")
x = model.addVars(DAYS, NFOODS, lb=0.0, name="x")           # servings
use = model.addVars(DAYS, NFOODS, vtype=GRB.BINARY, name="use")  # whether food used that day

# Objective: minimize total weekly cost
model.setObjective(gp.quicksum(costs[i] * x[d, i] for d in range(DAYS) for i in range(NFOODS)), GRB.MINIMIZE)

# Per-day nutrient bounds (re-use nutrient_bounds from notebook if available)
if 'nutrient_bounds' not in globals():
    raise RuntimeError("nutrient_bounds is not defined in notebook. Run earlier cells.")

# Add per-day nutrient constraints
for d in range(DAYS):
    for j in range(10):
        model.addConstr(
            gp.quicksum(nutrition_data[i][j] * x[d, i] for i in range(NFOODS)) >= nutrient_bounds[j][0],
            name=f"nutrient_{j}_min_day{d}"
        )
        model.addConstr(
            gp.quicksum(nutrition_data[i][j] * x[d, i] for i in range(NFOODS)) <= nutrient_bounds[j][1],
            name=f"nutrient_{j}_max_day{d}"
        )

    # minimum distinct items per day (via 'use' binaries)
    model.addConstr(gp.quicksum(use[d, i] for i in range(NFOODS)) >= MIN_DISTINCT_ITEMS_DAY,
                    name=f"min_distinct_day{d}")

    # link x and use: if use==1 then x >= MIN_SERVING_IF_USED and x <= BIG_M*use
    for i in range(NFOODS):
        model.addConstr(x[d, i] <= BIG_M * use[d, i], name=f"link_upper_d{d}_i{i}")
        model.addConstr(x[d, i] >= MIN_SERVING_IF_USED * use[d, i], name=f"link_lower_d{d}_i{i}")

    # optional per-day 30% caps (calories and protein)
    if ENFORCE_30_PCT_CAP:
        total_cal = gp.quicksum(nutrition_data[i][0] * x[d, i] for i in range(NFOODS))
        total_prot = gp.quicksum(nutrition_data[i][5] * x[d, i] for i in range(NFOODS))
        for i in range(NFOODS):
            model.addConstr(nutrition_data[i][0] * x[d, i] <= 0.3 * total_cal, name=f"cap_cal_d{d}_i{i}")
            model.addConstr(nutrition_data[i][5] * x[d, i] <= 0.3 * total_prot, name=f"cap_prot_d{d}_i{i}")

    # category rules (simple): Fruit/Vegetable at least 1 serving/day if they exist
    for group_name, indices in food_groups.items():
        if group_name.lower() in ("fruit", "fruits", "vegetable", "vegetables"):
            model.addConstr(gp.quicksum(x[d, i] for i in indices) >= 1.0, name=f"{group_name}_min1_day{d}")
            # soft realism: cap at 4 servings
            model.addConstr(gp.quicksum(x[d, i] for i in indices) <= 4.0, name=f"{group_name}_max4_day{d}")
        else:
            # require at least 0.5 servings from other groups (relaxed)
            model.addConstr(gp.quicksum(x[d, i] for i in indices) >= 0.5, name=f"{group_name}_min0.5_day{d}")

# Weekly repetition cap: each food appears at most MAX_REPEAT_PER_WEEK times
for i in range(NFOODS):
    model.addConstr(gp.quicksum(use[d, i] for d in range(DAYS)) <= MAX_REPEAT_PER_WEEK, name=f"max_repeat_week_i{i}")

# Solve
model.params.OutputFlag = 1
print("Solving Part D (5-day plan). ENFORCE_30_PCT_CAP =", ENFORCE_30_PCT_CAP)
model.optimize()

# Collect and export results
if model.status == GRB.OPTIMAL:
    week_rows = []
    daily_costs = []
    for d in range(DAYS):
        day_cost = 0.0
        for i in range(NFOODS):
            val = x[d, i].x
            if val is None:
                val = 0.0
            if val > 1e-6:
                cost_contrib = costs[i] * val
                day_cost += cost_contrib
                # determine category label if available
                category = None
                if 'food_group_labels' in globals() and food_group_labels is not None:
                    try:
                        category = food_group_labels[i]
                    except Exception:
                        category = "Unknown"
                else:
                    category = "Unknown"
                week_rows.append({
                    "Day": f"Day{d+1}",
                    "Food": foods[i],
                    "Servings": round(val, 3),
                    "Category": category,
                    "Cost": round(cost_contrib, 2)
                })
        daily_costs.append({"Day": f"Day{d+1}", "DailyCost": round(day_cost, 2)})

    df_week = pd.DataFrame(week_rows).sort_values(["Day","Food"])
    df_costs = pd.DataFrame(daily_costs)
    total_week_cost = df_costs['DailyCost'].sum()
    df_costs.loc[len(df_costs)] = {"Day":"TotalWeek","DailyCost": round(total_week_cost, 2)}

    # Export to Excel
    with pd.ExcelWriter(OUTPUT_XLSX, engine="openpyxl") as writer:
        df_week.to_excel(writer, sheet_name="Week_Plan", index=False)
        df_costs.to_excel(writer, sheet_name="Costs", index=False)

    print(f"Optimal weekly cost: ${model.objVal:.2f}")
    print(f"Results exported to {OUTPUT_XLSX}")
    print("\nQuick summary (top contributors per day):")
    for d in range(DAYS):
        day_df = df_week[df_week['Day']==f"Day{d+1}"].sort_values("Cost", ascending=False).head(5)
        print(f" Day {d+1}: {len(day_df)} items selected, top items:")
        for _, r in day_df.iterrows():
            print(f"  - {r['Food']}: {r['Servings']} servings (${r['Cost']})")
else:
    print(f"Optimization did not succeed. Status = {model.status}")
    if model.status == GRB.INFEASIBLE:
        print("Model infeasible. Trying to write IIS for diagnosis...")
        try:
            model.computeIIS()
            iis_path = "partd_moves.iis"
            model.write(iis_path)
            print(f"IIS written to {iis_path}. Inspect the file to find conflicting constraints.")
        except Exception as e:
            print("Failed to compute/write IIS:", e)
    elif model.status == GRB.UNBOUNDED:
        print("Model unbounded. Check objective / variable bounds.")
```

```python
# Auto-relax helper: iteratively loosen realism constraints until feasible
# This cell will try configurations in order and stop on first feasible solution.

import copy
import time
from gurobipy import GRB

# Base parameters (read from current notebook state)
BASE_MIN_DISTINCT = MIN_DISTINCT_ITEMS_DAY
BASE_MIN_SERVING = MIN_SERVING_IF_USED
BASE_MAX_REPEAT = MAX_REPEAT_PER_WEEK
BASE_CATEGORY_UPPER = 4.0  # current cap used in model

relax_configs = []
# Strategy: progressively relax distinct items, min serving, max repeat, remove category upper caps, relax nutrient mins by small percentages
for distinct in [max(1, BASE_MIN_DISTINCT-2), max(1, BASE_MIN_DISTINCT-1), BASE_MIN_DISTINCT]:
    for min_serv in [max(0.1, BASE_MIN_SERVING/2), BASE_MIN_SERVING, BASE_MIN_SERVING*1.5]:
        for max_rep in [BASE_MAX_REPEAT, BASE_MAX_REPEAT+1, BASE_MAX_REPEAT+2]:
            for remove_cat_upper in [False, True]:
                for nutrient_relax_pct in [0.0, 0.05, 0.10]:  # relax nutrient mins by up to 10%
                    relax_configs.append({
                        'MIN_DISTINCT_ITEMS_DAY': distinct,
                        'MIN_SERVING_IF_USED': min_serv,
                        'MAX_REPEAT_PER_WEEK': max_rep,
                        'REMOVE_CATEGORY_UPPER': remove_cat_upper,
                        'NUTRIENT_MIN_RELAX_PCT': nutrient_relax_pct
                    })

print(f"Prepared {len(relax_configs)} relaxation configurations; beginning attempts...")

def try_config(cfg, time_limit=10):
    # Rebuild the model similar to Part D but with cfg overrides
    DAYS_local = DAYS
    NFOODS_local = NFOODS
    model_r = gp.Model()
    x = model_r.addVars(DAYS_local, NFOODS_local, lb=0.0, name="x")
    use = model_r.addVars(DAYS_local, NFOODS_local, vtype=GRB.BINARY, name="use")
    model_r.setObjective(gp.quicksum(costs[i] * x[d, i] for d in range(DAYS_local) for i in range(NFOODS_local)), GRB.MINIMIZE)

    # apply nutrient bounds with min relaxation
    for d in range(DAYS_local):
        for j in range(10):
            minb = nutrient_bounds[j][0] * (1.0 - cfg['NUTRIENT_MIN_RELAX_PCT'])
            maxb = nutrient_bounds[j][1]
            model_r.addConstr(gp.quicksum(nutrition_data[i][j] * x[d, i] for i in range(NFOODS_local)) >= minb)
            model_r.addConstr(gp.quicksum(nutrition_data[i][j] * x[d, i] for i in range(NFOODS_local)) <= maxb)

        model_r.addConstr(gp.quicksum(use[d, i] for i in range(NFOODS_local)) >= cfg['MIN_DISTINCT_ITEMS_DAY'])

        for i in range(NFOODS_local):
            model_r.addConstr(x[d, i] <= BIG_M * use[d, i])
            model_r.addConstr(x[d, i] >= cfg['MIN_SERVING_IF_USED'] * use[d, i])

        # category rules
        for group_name, indices in food_groups.items():
            if group_name.lower() in ("fruit", "fruits", "vegetable", "vegetables"):
                model_r.addConstr(gp.quicksum(x[d, i] for i in indices) >= 1.0)
                if not cfg['REMOVE_CATEGORY_UPPER']:
                    model_r.addConstr(gp.quicksum(x[d, i] for i in indices) <= BASE_CATEGORY_UPPER)
            else:
                model_r.addConstr(gp.quicksum(x[d, i] for i in indices) >= 0.5)

    for i in range(NFOODS_local):
        model_r.addConstr(gp.quicksum(use[d, i] for d in range(DAYS_local)) <= cfg['MAX_REPEAT_PER_WEEK'])

    model_r.params.OutputFlag = 0
    model_r.params.TimeLimit = time_limit
    try:
        model_r.optimize()
    except gp.GurobiError as e:
        return {'status': 'error', 'error': str(e)}

    return {'status': model_r.status, 'model': model_r if model_r.status == GRB.OPTIMAL else None}

# Try configurations sequentially
feasible_result = None
attempt = 0
start_time = time.time()
for cfg in relax_configs:
    attempt += 1
    print(f"Attempt {attempt}: {cfg}")
    res = try_config(cfg, time_limit=6)
    if res.get('status') == GRB.OPTIMAL:
        feasible_result = {'cfg': cfg, 'model': res['model']}
        print("Found feasible configuration:", cfg)
        break
    elif res.get('status') == GRB.INFEASIBLE:
        print(" -> infeasible")
    elif res.get('status') == GRB.TIME_LIMIT:
        print(" -> time limit reached (no feasible found within limit)")
    else:
        print(" -> status:", res.get('status'))

end_time = time.time()
print(f"Auto-relax attempts completed in {end_time - start_time:.2f}s")

if feasible_result is not None:
    # Extract solution and export similar to Part D
    model_ok = feasible_result['model']
    cfg = feasible_result['cfg']
    print('Feasible config:', cfg)
    # Rebuild solution extraction using x variable names (they exist in model_ok)
    week_rows = []
    daily_costs = []
    # retrieve variable dicts
    x_vars = {v.VarName: v for v in model_ok.getVars() if v.VarName.startswith('x[')}
    # parse names like x[0,1]
    import re
    def parse_x_name(name):
        m = re.match(r"x\[(\d+),(\d+)\]", name)
        if m:
            return int(m.group(1)), int(m.group(2))
        return None

    for vname, v in x_vars.items():
        parsed = parse_x_name(vname)
        if parsed:
            d, i = parsed
            val = v.X
            if val > 1e-6:
                cost_contrib = costs[i] * val
                week_rows.append({'Day': f'Day{d+1}', 'Food': foods[i], 'Servings': round(val,3), 'Category': (food_group_labels[i] if 'food_group_labels' in globals() and food_group_labels is not None else 'Unknown'), 'Cost': round(cost_contrib,2)})
    # compute daily costs
    df_week = pd.DataFrame(week_rows)
    if not df_week.empty:
        df_costs = df_week.groupby('Day')['Cost'].sum().reset_index().rename(columns={'Cost':'DailyCost'})
        total_week_cost = df_costs['DailyCost'].sum()
        df_costs.loc[len(df_costs)] = {'Day':'TotalWeek', 'DailyCost': round(total_week_cost,2)}
    else:
        df_costs = pd.DataFrame(columns=['Day','DailyCost'])

    with pd.ExcelWriter(OUTPUT_XLSX, engine='openpyxl') as writer:
        df_week.to_excel(writer, sheet_name='Week_Plan', index=False)
        df_costs.to_excel(writer, sheet_name='Costs', index=False)

    print(f"Feasible solution exported to {OUTPUT_XLSX}")
else:
    print('No feasible configuration found within the tested relaxations.')

```
